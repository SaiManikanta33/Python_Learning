Multithreaded Port Scanner

Use only on systems you own or have permission to scan.

```python
import socket
from concurrent.futures import ThreadPoolExecutor

host = "192.168.2.11"

def scan(port):
    sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    sock.settimeout(1)

    if sock.connect_ex((host, port)) == 0:
        print(f"Port {port}: OPEN")

    sock.close()

ports = range(20, 101)

with ThreadPoolExecutor(max_workers=20) as executor:
    executor.map(scan, ports)
    
```

Concurrent Log Analyzer

Create three log files:

auth.log
system.log
web.log

Each contains sample log entries.

```python
from concurrent.futures import ThreadPoolExecutor

files = ["auth.log", "system.log", "web.log"]

def read_log(file):
    with open(file, "r") as f:
        print(f"\nReading {file}")
        print(f.read())

with ThreadPoolExecutor(max_workers=3) as executor:
    executor.map(read_log, files)
```
