Exercise 1 — Extract Numbers

Given:

text = "There are 25 users and 10 servers."

Extract:

25
10

```python
import re
text = "There are 25 users and 10 servers."
print(re.findall(r"\d+",text))
```

Exercise 2 — Extract Emails

Given:

text = """
admin@gmail.com
security@example.com
test@company.org
"""

Extract all emails.

```python
import re
text = """
admin@gmail.com
security@example.com
test@company.org
"""
pattern = r"[\w.-]+@[\w.-]+\.\w+"
emails = re.findall(pattern,text)
print(emails)
```

Exercise 3 — Extract IP Addresses

Given:

text = """
192.168.1.1
10.0.0.15
172.16.2.20
"""

Extract all IP-shaped values.

```python
import re
text = """
192.168.1.1
10.0.0.5
172.16.2.20
"""
pattern = r"\b(?:\d{1,3}\.){3}\d{1,3}\b"
res = re.findall(pattern,text)
print(res)
```

Exercise 4 — Redact Passwords

Given:

text = "username=admin password=hello123"

Convert it into:

username=admin password=REDACTED

```python
import re
text = "username=admin password=hello123"
res=re.sub(
    r"password=\S+",
    "password=Pass123",
    text
)
print(res)
```

Exercise 5 — Extract User + IP

Given:

User: mani IP: 192.168.1.50

Extract:

Username: mani
IP: 192.168.1.50

Use capturing groups.

```python
import re
text = "User: mani IP: 192.168.1.50"
pattern = r"User:\s(\w+)\sIP:\s([\d.]+)"
res = re.search(pattern,text)
print(res.group(1))
print(res.group(2))
```
