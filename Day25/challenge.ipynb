```python
import sqlite3
connect = sqlite3.connect("college.db")
cursor = connect.cursor()
cursor.execute("""
               create table if not exists college1(
                   id integer primary key,
                   name text,
                   branch text,
                   cgpa real
                   )
                   """)

cursor.execute("insert into college1 values(?,?,?,?)",
               (1,"Sai","Cyber Security",9.1))
connect.commit()
cursor.execute("select * from college1")
for student in cursor.fetchall():
    print(student)
```
