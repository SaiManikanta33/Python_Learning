Exercise 1 — GET Request

Send a GET request to:

https://api.github.com

Print:

-- Status code
-- Response headers
-- Response JSON

```python
import requests
url="https://api.github.com"

response = requests.get(url)

print(response.status_code)
print(response.headers)
print(response.text)
 
```

Exercise 2 — Status Checker

Write a program that accepts a URL:

    url = input("Enter URL: ")

Then prints:

    Website Status: 200

Handle connection errors.

```python
import requests
url = input("Enter URL:")
response = requests.get(url)
if response.status_code == 200:
    print(f"Website Status: {response.status_code}")
else:
    print("Invalid ")
```

Exercise 3 — JSON

Create:

data = {
    "username": "admin",
    "role": "SOC Analyst",
    "active": True
}

Convert it into JSON and print it.

```python
import json
data = {
    "username": "admin",
    "role": "SOC Analyst",
    "active": True
}
json_data = json.dumps(data)
print(json_data)
```

Exercise 4 — Query Parameters

Create a GET request using:

params = {
    "page": 1,
    "limit": 10
}

Print the final URL.

```python
import requests
params = {
    "page":1,
    "limit":10
}
response = requests.Request(
    "GET",
    "https://api.github.com",
    params=params
).prepare()
print(response.url)
```

Exercise 5 — Headers

Create a request with:

headers = {
    "User-Agent": "Python-Security-Lab"
}

Print the response status.

```python
import requests
url = "https://api.github.com"
headers = {
    "User-Agent":"Python-Security-Lab"
}
res = requests.get(url,headers=headers)
print(f"Response Code : {res.status_code}")
```
