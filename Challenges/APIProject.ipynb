Threat Intelligence API Client

Today you're going to combine:

Python + APIs + Cybersecurity

Build a program that accepts an IP address:

--Enter IP address: 8.8.8.8

Then sends it to a threat-intelligence or IP-information API that you are authorized to use.

Your program should display information such as:

--IP Address: 8.8.8.8
--Country: ...
--Organization: ...
--ASN: ...
--Reputation: ...

Important

Use a legitimate API and follow its terms, rate limits, and authentication requirements.

Don't expose your API key in GitHub.
