Exercise 1

Create two threads.

One prints:

Learning Python

The other prints:

Learning Cyber Security

```python
import threading

def greet():
    print("Learning Python")
    print("Learning Cyber Security")

thread = threading.Thread(target=greet)

thread.start()

thread.join()
```

Exercise 2

Create five threads.

Each prints its thread number.

```python
import threading
import time

def worker(name):
    print(f"{name} started")
    time.sleep(2)
    print(f"{name} finished")

threads = []

for i in range(5):
    t = threading.Thread(target=worker, args=(f"Thread-{i+1}",))
    threads.append(t)
    t.start()

for t in threads:
    t.join()

print("All threads completed.")
```
