Exercise 1

Create a generator that yields numbers from 1 to 20.

```python
def num(limit):
    for n in range(limit):
        yield n
for value in num(21):
    print(value)
```

Exercise 2

Create a generator that yields odd numbers only.

```python
def odd_numbers(limit):
    for n in range(limit):
        if n % 2!=0:
            yield n
for num in odd_numbers(20):
    print(num)
```

Exercise 3

Create a generator expression that returns cubes of numbers from 1 to 10.

```python
def cubes(limit):
    for n in range(limit):
        yield n**n
for num in cubes(10):
    print(num)
```

Exercise 5

Create a generator that yields Fibonacci numbers.

```python
def fib(limit):
    a,b=0,1
    for _ in range(limit):
        yield a
        a,b =b,a+b
for num in fib(10):
    print(num)
```
