```python
import re
with open("security.log","r")as file:
    logs = file.read()
    
errors = re.findall(r"ERROR.*",logs)
warnings = re.findall(r"WARNING.*",logs)
infos = re.findall(r"INFO.*",logs)

print("Errors:")
count = 0
for error in errors:
    count += 1
    print(error)
print("count of Errors Are: ",count)
    
print("\n Warning:")
for warning in warnings:
    print(warning)
print("Count Of Warnind Are: ",count)

print("INFO \n")
for info in infos:
    print(info)
print("Count of Infos Are: ",count)


total_logs = len(errors) + len(warnings) + len(infos)
if total_logs > 0:
    print(f"Error percentage :{len(errors)/total_logs * 100:.2f}%")
    print(f"Warning percentage:{len(warnings)/total_logs * 100:.2f}%")
    print(f"INFO percentage:{len(infos)/total_logs * 100:.2f}%")
else:
    print("Logs Not Found")
with open("report.txt","w")as file:
    file.write(f"Error: \n")
    file.write("\n".join(errors)+"\n")
    file.write(f"Count of Errors:{len(errors)}\n\n")
    file.write(f"Error percentage :{len(errors)/total_logs * 100:.2f}%")
    
    file.write("Warnings:\n")
    file.write("\n".join(warnings) + "\n")
    file.write(f"Count of Warnings: {len(warnings)}\n\n")
    
    file.write("Warnings:\n")
    file.write("\n".join(warnings) + "\n")
    file.write(f"Count of Warnings: {len(warnings)}\n\n")
```
