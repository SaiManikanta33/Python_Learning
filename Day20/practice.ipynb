Exercise 5

Scan ports 20–30 on 127.0.0.1 and display which are open.

```python
import socket
host = "127.0.0.1"
for port in range(20,31):
    sock = socket.socket(socket.AF_INET,socket.SOCK_STREAM)
    sock.settimeout(0.8)
    res = sock.connect_ex((host,port))
    if res == 0:
        print(f"Port {port} is open")
    else:
        print(f"Port {port } in close")
    sock.close()
```
