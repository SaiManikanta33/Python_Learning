```python
import logging

# Main log: every login event
logging.basicConfig(
    filename="security_events.log",
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s",
    force=True
)

# Failed-login log: only failed attempts
failed_logger = logging.getLogger("failed_logins")
failed_logger.setLevel(logging.WARNING)
failed_logger.propagate = False

failed_handler = logging.FileHandler("failed_events.log")
failed_handler.setFormatter(
    logging.Formatter("%(asctime)s | %(levelname)s | %(message)s")
)

# Prevent duplicate logs when rerunning a notebook cell
failed_logger.handlers.clear()
failed_logger.addHandler(failed_handler)

username = input("Username: ")
status = input("Login Status (Success/Failed): ").strip().title()
ip = input("Enter IP Address: ")
device_name = input("Enter Device Name: ")

event = (
    f"User={username} Status={status} "
    f"IP={ip} DeviceName={device_name}"
)

# Saves every attempt, including the login time from %(asctime)s
logging.info(event)

# Saves failed attempts separately
if status == "Failed":
    failed_logger.warning(event)

# Count all failed attempts saved so far
try:
    with open("security_events.log", "r", encoding="utf-8") as file:
        failed_logins = sum(
            1 for line in file if "Status=Failed" in line
        )
except FileNotFoundError:
    failed_logins = 0

print("Event Logged Successfully")
print("Total Failed Logins:", failed_logins)
```

Write a script that:

Logs every user with more than 3 failed attempts.

Logs users with more than 7 failed attempts as CRITICAL.

Example:

WARNING User admin has 5 failed logins

CRITICAL User sai has 8 failed logins

```python
import logging
logging.basicConfig(filename="security_events.log", level=logging.INFO, format="%(asctime)s | %(levelname)s | %(message)s")
failed_attempts={"admin":5,
                 "guest":2,
                 "Sai":8
                 }
for username,attempts in failed_attempts.items():
    if attempts > 7:
        logging.critical(f"User {username} has {attempts} failed login")
    elif attempts > 3:
        logging.warning(f"User {username} has {attempts} failed logins")
        
```
