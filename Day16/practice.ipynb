Exercise 2

Create a decorator that prints:

Starting...
Ending...

around any function.

```python
def A1(sai):
    def wrapper():
        print("Starting...")
        sai()
        print("Ending...")
    return wrapper
@A1
def greet():
    print("Helloooooo!")
greet()
```

Exercise 3

Create a decorator that counts how many times a function is called.

Example:

Call #1
Call #2
Call #3

```python
def call(func):
    func.call =  0
    def wrapper(*args,**kwargs):
        func.call += 1
        print(f"Call #{func.call}")
        return func(*args,**kwargs)
    return wrapper
@call
def greet():
    print("Function executed")
greet()
greet()
greet()
greet()
greet()
```

Exercise 4

Create a custom context manager named LogSession.

Output:

Log Started
Reading Logs...
Log Closed

```python
class DBConnect:
    def __enter__(self):
        print("Log Started")
        return self
    def __exit__(self,exc_type,exc_val,exc_tb):
        print("Log Colsed")
with DBConnect():
    print("Reading Logs...")
```

Exercise 5

Write a decorator that validates whether an age is at least 18 before allowing access.

```python
def sai(age):
    def wrapper():
        print("Validating Age...")
        res =  age()
        print("Age is Valid")
        return res
    return wrapper
@sai
def greet():
    if age >= 18:
        return "Allowing the access"
    else:
        return "Age is atleast 18"
age = 19
print(greet())
```
