```python
import datetime
def log_execution(func):
    def wrapper(*args,**kwargs):
        print(f"[{datetime.datetime.now()}] Executing {func.__name__}")
        return func(*args,**kwargs)
    return wrapper
@log_execution
def login(username):
    print(f"{username} logged in")
login("Sai")
login("Manikanta")
```

Login Authentication Decorator

Create a decorator that checks if a user is authenticated before allowing access.

```python
authenticated = True
failed_attempts = 0
MAX_ATTEMPTS = 3
def require_login(func):
    def wrapper():
        global failed_attempts
        if failed_attempts >= MAX_ATTEMPTS:
            print("Account locked.")
            return 
        username = input("Enter Username:")
        
        if username == "admin" and authenticated:
            print("Login successful.")
            failed_attempts = 0
            return func(username)
        
        if failed_attempts >= MAX_ATTEMPTS:
            print("Account locked after 3 failed attempts.")

    return wrapper
@require_login
def access_dashboard(name):
    print(f"Welcome { name} to the SOC Dashboard")
access_dashboard()
access_dashboard()
access_dashboard()
```
