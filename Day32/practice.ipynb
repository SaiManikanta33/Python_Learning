Exercise 1

Create an async function that prints your name after waiting 2 seconds.

```python
import asyncio
async def msg(name):
    print("Name is printing")
    await asyncio.sleep(2)
    print(f"{name} Printed after 2 seconds")
    
await (msg("Sai Manikanta"))
```

You’re running this in Jupyter/IPython, which already has an active event loop. 
Use "await" directly instead of asyncio.run():

Exercise 2

Run three async tasks simultaneously using asyncio.gather().

```python
import asyncio
async def task(file):
    print(f"Downloading {file}")
    await asyncio.sleep(3)
    print(f"{file} downloading")
    
async def main():
    task1 = asyncio.create_task(task("Report.pdf"))
    task2 = asyncio.create_task(task("Logs.csv"))
    task3 = asyncio.create_task(task("Notes.txt"))
    
    
    await task1
    await task2
    await task3
    
await (main())
```

Exercise 3

Create an async countdown timer.

```python
import asyncio
async def timer(start,stop):
    while start <= stop:
        print(start)
        await asyncio.sleep(0.5)
        start += 1
    print("Time's up! ⏰")
await timer(1,20)    
```

Exercise 4

Use aiohttp to fetch two websites and print their HTTP status codes.

```python
import aiohttp
import asyncio
async def get(url):
    async with aiohttp.ClientSession()as session:
        async with session.get(url)as response:
            print(url,response.status)

async def main():
    await asyncio.gather(
        get("https://python.org")
    )
    
await (main())
```

Exercise 5

Measure the difference between synchronous and asynchronous execution using time.perf_counter().

Synchronous Version

```python
import asyncio
import time
def task(name):
    print(f"Starting {name}")
    time.sleep(2)
    print(f"Finished {name}")
    
start = time.perf_counter()
task("Task 1")
task("Task 2")
task("Task 3")
end = time.perf_counter()
print(f"\n Synchromous time:{end-start:.2f} seconds")

print("-"*5,"Asynchrounous Version","-"*5)

async def tasks(name):
    print(f"Starting {name}")
    await asyncio.sleep(2)
    print(f"Finished {name}")
    
async def main():
    start = time.perf_counter()
    await asyncio.gather(
        tasks("Task 1"),
        tasks("Task 2"),
        tasks("Task 3")
    )

    end = time.perf_counter()
    print(f"\n Asynchronous time: {end - start:.2f} seconds")
await main()
```
