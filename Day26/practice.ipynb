Exercise 1

Create a log file named:

student_app.log

Log:

Application started
User logged in
User logged out

```python
import logging
logging.basicConfig(filename="student_app.log",level=logging.INFO,force=True)
logging.info("Application Started")
logging.info("User Logged in")
logging.info("User logged out")
```

```python
import logging
logging.basicConfig(filename="student_app.log",level=logging.DEBUG,force=True)
logging.debug("Debug message")
logging.info("Application successfully started")
logging.warning("Low power")
logging.error("Integration failed")
logging.critical("Application crashed")
```

Exercise 3

Create a custom log format that includes:

Time
Log level
Message

```python
import logging
logging.basicConfig(filename="student_app.log",level=logging.INFO,force=True,format="%(asctime)s | %(levelname)s | %(message)s")
logging.debug("Dubug Seccessfully")
logging.info("Application started")
```

Exercise 5

Create a rotating log file that automatically rotates after 2 KB

```python
import logging
from logging.handlers import RotatingFileHandler

handler = RotatingFileHandler(
    "security.log",
    maxBytes=5,
    backupCount=3,
    
)

logging.basicConfig(
    handlers=[handler],
    level=logging.INFO,
    force=True,
    format="%(asctime)s | %(levelname)s | %(message)s"
)

logging.info("Security Event")
```
