Exercise 3

Write a program that logs:

--Program started
--User logged in
--Program finished

to a file.

```python
import logging
logging.basicConfig(
    filename="logs.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    
)
logging.info("Program started")
logging.warning("Warning message")
logging.error("Error occured")
```
