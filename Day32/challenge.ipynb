🧩 Mini Project
Website Status Checker

```python
import asyncio
import aiohttp
import time
import csv
websites = [
    "https://instagram.com",
    "https://python.org",
    "https://github.com"
]

async def check_status(url):
    async with aiohttp.ClientSession()as session:
        async with session.get(url)as response:
            print(f"{url}:{response.status}")
            return [url,response.status]
            
async def main():
    start = time.perf_counter()
    async with aiohttp.ClientSession()as session:
        results = await asyncio.gather(*(check_status(url) for url in websites))
    end = time.perf_counter()
    time_taken=end - start
    print(f"The response is taken : {time_taken:.2f} seconds")
    return results,time_taken

results, time_taken = await main()

with open("status.csv","w",newline="")as file:
    writer = csv.writer(file)
    writer.writerow(["Website","Status Code"])
    writer.writerows(results)
    writer.writerow([])
    writer.writerow(["Total Response Time(seconds) : "f"{time_taken:.2f}"])
    
print("Results Saved to CSV file")
```
