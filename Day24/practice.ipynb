Exercise 3

Sort students by CGPA in descending order.

```python

sorted_data = data.sort_values(by="CGPA",ascending=False)
print(sorted_data)
```

Exercise 4

Filter students whose CGPA is greater than or equal to 8.5

```python
filter_data = data[data["CGPA"]>=8.5]
print(filter_data)
```
