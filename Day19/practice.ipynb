Exercise 1

Extract all numbers from:

text = "Ports 22, 80, 443, 3389"

Expected output:

['22', '80', '443', '3389']

```python
import re
txt = "Ports 20,80,443,3389"
res = re.findall(r"\d+",txt)
print(res)
```

Exercise 2

Validate these emails:

admin@test.com
invalid-email
cyber@company.org

Print whether each one is valid or invalid.

```python
import re
txt =["admin@test.com","invalid-email","cyber@company.org"]
pattern = r"^[\w\.-]+@[\w\.-]+\.\w+$"
for email in txt:
    if re.match(pattern,email):
        print("Matched: ",email)
    else:
        print("Not matched: ",email)
```

Exercise 3

Extract all IPv4-like addresses from:

"""
192.168.1.1
10.10.10.5
172.16.1.100
"""

```python
import re
ip = """
192.168.1.1
10.10.10.5
172.16.1.100
"""
res = re.findall(r"\b(?:\d{1,3}\.){3}\d{1,3}\b",ip)
print(res)
```

It retuns any number have 3 digits
So overcome by using this expression
pattern = r"\b(?:(?:25[0-5]|2[0-4]\d|1\d{2}|[1-9]?\d)\.){3}(?:25[0-5]|2[0-4]\d|1\d{2}|[1-9]?\d)\b"

```python
import re
ip = """
192.168.1.1
600.10.10.5
172.999.1.100
"""
res = re.findall(r"\b(?:(?:25[0-5]|2[0-4]\d|1\d{2}|[1-9]?\d)\.){3}(?:25[0-5]|2[0-4]\d|1\d{2}|[1-9]?\d)\b",ip)
print(res)
```

Exercise 5

Extract all URLs from:

https://google.com
http://github.com
https://openai.com

```python
import re
mails="https://google.com or http://github.com or https://openai.com"
url = re.findall(r"http?://[^\s]+",mails)
print(url)
```
