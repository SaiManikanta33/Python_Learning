REST API User Viewer

```python
import requests
import json
url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)
response.raise_for_status
users = response.json()
with open("users.json","w")as file:
    json.dump(users,file,indent=4)
print("Users saved to users.json")
sorted_users = sorted(users,key=lambda user:user["name"].lower())

print("\n Users sorted alphabetically:")
for user in sorted_users:
    print(f"{user["name"]} - {user["email"]}")
    
    
email_to_find=input("\n Enter an Enail:").strip().lower()
found_user = None
for user in users:
    if user["email"].lower() == email_to_find:
        found_user = user
        break
if found_user:
    print("\nUser found:")
    print("Name:", found_user["name"])
    print("Email:", found_user["email"])
    print("Company:", found_user["company"]["name"])
else:
    print("\nNo user found with that email address.")
```
