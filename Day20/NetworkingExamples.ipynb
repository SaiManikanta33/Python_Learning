```python

    #Creating a TCP Server

import socket
server = socket.socket(socket.AF_INET,socket.SOCK_STREAM)
server.bind(("192.168.2.2",5000))
server.listen(1)
print("Wating for connection...")
client,address = server.accept()
print("Connected:",address)
client.send(b"Welcome to python Server!")
client.close()
server.close()
```

```python

    # Createing a TCP Client

import socket
client = socket.socket(socket.AF_INET,socket.SOCK_STREAM)
client.connect(("127.0.0.1",5000))
message = client.recv(1024).decode()
client.send("Hello Client".encode())
messages = client.recv(1024).decode()
print(message)
print(messages)
client.close()
```

```python
import socket

server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
server.bind(("127.0.0.1", 5000))
server.listen(1)

print("Server listening on port 5000...")
connection, address = server.accept()
print("Connected by:", address)

connection.sendall(b"Welcome, client!")
message = connection.recv(1024)
print("Client says:", message.decode())

connection.sendall(b"Hello from server!")
connection.close()
server.close()
```

```python
        #creating a simple port scanner
        
import socket
host = "127.0.0.1"
ports=[22,80,443,8080]
for port in ports:
    sock = socket.socket(socket.AF_INET,socket.SOCK_STREAM)
    sock.settimeout(1)
    result = sock.connect_ex((host,port))
    if result == 0:
        print(f"Port {port} is open")
    else:
        print(f"Port {port} is CLOSED")
    sock.close()
```

```python
import socket

host = "127.0.0.1"  # Replace only with an IP you are authorized to scan

for port in range(1, 65536):
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as sock:
        sock.settimeout(0.1)  # seconds

        if sock.connect_ex((host, port)) == 0:
            print(f"Port {port} is open")
```

```python
import socket
host = "127.0.0.1"
for port in range(20,30):
    sock = socket.socket(socket.AF_INET,socket.SOCK_STREAM)
    sock.settimeout(2)
    result = sock.connect_ex((host, port))

    if result == 0:
        print(f"Port {port} is OPEN")
    else:
        print(f"Port {port} is CLOSED")

    sock.close()
```
