```python
import logging
logging.basicConfig(
    filename="login.log",
    level=logging.INFO,
    format="%(asctime)s - %(message)s"
)
failed = 0
def process_login(username,input_pass,correct_pass,user_ip="1920168.1.45"):
    global failed
    if input_pass == correct_pass:
        logging.info(f"Login Success - User:{username} | IP:{user_ip}")
        print("🎉 Login successful! Welcome dashboard loading...")
        return True
    else:
        failed += 1
        logging.warning(
            f"Login Failed - User: {username} | IP: {user_ip}"
            f"Attempt Count: {failed}"
        )
        print("❌ Invalid password. Access denied.")
        return False
if __name__ == "__main__":
    print("--- Starting Login Security Test ---")
    
    # Target system credentials
    real_pass = "SecurePassword123"
    
    # Scenario 1: First wrong password attempt
    process_login(username="sai_manikanta", input_pass="WrongPassword", correct_pass=real_pass)
    
    # Scenario 2: Second wrong password attempt (Increments counter)
    process_login(username="sai_manikanta", input_pass="GuessingPassword", correct_pass=real_pass)
    
    # Scenario 3: Successful login attempt
    process_login(username="sai_manikanta", input_pass="SecurePassword123", correct_pass=real_pass)
    
    print("\n--- Session Complete. Check logs/app.log for details! ---")
```
