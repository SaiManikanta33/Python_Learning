Exercise 1

Create two threads.

Each thread prints numbers from 1–5.

```python
import threading
import time
def greet():
    for i in range(1,6):
        print(i)
        time.sleep(2)
        
thread1 = threading.Thread(target=greet)
thread1.start()

thread1.join()
```

Exercise 2

Pass different names to two threads.

```python
import threading
import time
def greet(name):
    for i in range(5):
        print(f"{name}:{i}")
        time.sleep(2)
        
t1=threading.Thread(target=greet,args=("Thread A",))
t2=threading.Thread(target=greet,args=("Thread B",))

t1.start()
t2.start()

t1.join()
t2.join()
```

Exercise 3

Create five threads that print the current time.

```python
import threading
from datetime import datetime
def show_time():
    for i in range(1,6):
        print(datetime.now(),i)
        time.sleep(2)
t2  = threading.Thread(target=show_time)
t2.start()
t2.join()
```

Exercise 4

Use ThreadPoolExecutor to calculate squares of numbers.

```python
from concurrent.futures import ThreadPoolExecutor
def show(number):
    time.sleep(1)
    return number * number
with ThreadPoolExecutor(max_workers=5)as exe:
    result = exe.map(show,[1,2,5,4])
print(list(result))
```

Exercise 5

Use a lock to safely increment a shared counter.

```python
import threading
count = 0
lock = threading.Lock()
def increment():
    global count
    with lock:
        count += 1
threads = []
for _ in range(200):
    thread = threading.Thread(target=increment)
    threads.append(thread)
    thread.start()
for thread in threads:
    thread.join()
print(count)
```
