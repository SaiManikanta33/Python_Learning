Multi-threaded File Reader

Create three text files.

file1.txt
file2.txt
file3.txt

```python
import threading
import os

report_lock = threading.Lock()

open("report.txt","w").close()
def read_file(filename):
    with open(filename, "r") as file:
        content = file.read()
    
    size = os.path.getsize(filename)
    print(f"\n{filename} - File Size:{size} bytes")

    print(content)

    word_count = len(content.split())
    print("Count of Words:", word_count)
    
    print(f"Count of lines :",len(content.splitlines()))
    
    report=(f"\nFile:{filename}\n"
            f"File Size:{size} bytes \n"
            f"Count of words:{word_count}"
            f"Count of lines:{len(content.splitlines())}")
    
    print(report)
    with report_lock:
        with open("report.txt","a")as file:
            file.write(report)

files = ["file1.txt", "file2.txt", "file3.txt"]
threads = []

for filename in files:
    thread = threading.Thread(target=read_file, args=(filename,))
    threads.append(thread)
    thread.start()

for thread in threads:
    thread.join()
```

Multi-threaded Port Scanner (Educational Example)

⚠️ Only scan systems you own or have explicit permission to test.

Challenge

Enhance it to:

--Scan ports 1–1024.
--Save open ports to a file.
--Measure scan time.
--Add a progress indicator.

```python
import socket
import threading
import time
from concurrent.futures import ThreadPoolExecutor, as_completed

target = "192.168.2.9"
ports = range(1, 1024)

open_ports = []
lock = threading.Lock()
completed = 0
total_ports = len(ports)

def scan(port):
    try:
        with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as sock:
            sock.settimeout(1)

            if sock.connect_ex((target, port)) == 0:
                with lock:
                    open_ports.append(port)
                print(f"\nPort {port} is OPEN")

    except socket.error:
        pass

    finally:
        global completed
        with lock:
            completed += 1
            print(
                f"\rProgress: {completed}/{total_ports} ports scanned",
                end="",
                flush=True
            )

start_time = time.perf_counter()

with ThreadPoolExecutor(max_workers=20) as executor:
    futures = [executor.submit(scan, port) for port in ports]

    for future in as_completed(futures):
        future.result()

elapsed_time = time.perf_counter() - start_time

with open("open_ports.txt", "w") as file:
    for port in sorted(open_ports):
        file.write(f"Port {port} is OPEN\n")

print("\n\nOpen ports saved to open_ports.txt")
print(f"Scan completed in {elapsed_time:.2f} seconds")
```
