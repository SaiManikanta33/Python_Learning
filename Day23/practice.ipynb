Exercise 1

Fetch the HTML from:

https://example.com

Print:

Status code
Page title

```python
import requests
from bs4 import BeautifulSoup
url = "https://google.com"
response = requests.get(url)
soup = BeautifulSoup(response.text,"html.parser")
print(response.status_code)
print(soup.title.text)
```

Exercise 2

Extract all hyperlinks (<a> tags).

```python
import requests
from bs4 import BeautifulSoup
url = "https://google.com"
response = requests.get(url)
soup = BeautifulSoup(response.text,"html.parser")
for link in soup.find_all("a"):
    print(link.get("href"))
```

Exercise 3

Extract all headings (<h1> and <h2>).

```python
import requests
from bs4 import BeautifulSoup
url = "https://www.raghuenggcollege.com/"
response = requests.get(url)
soup = BeautifulSoup(response.text,"html.parser")
for link in soup.find_all("h1"):
    print(link.get_text(strip=True))
for link in soup.find_all("h3"):
    print(link.get_text(strip=True))
```

Exercise 5

Save the page title and all links to a text file named:

page_info.txt

```python
import requests
from bs4 import BeautifulSoup
url = "https://www.raghuenggcollege.com/"
response = requests.get(url)
soup = BeautifulSoup(response.text,"html.parser")
print(soup.title.text)
for link in soup.find_all("p"):
    print(link.get_text(strip=True))
with open("page_info.txt","w")as file:
    print(file.write(f"{link.get_text(strip=True)}"))
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://example.com"

response = requests.get(url, timeout=10)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

print("Title:", soup.title.get_text(strip=True))

print("\nLinks:")

for link in soup.find_all("a"):
    href = link.get("href")
    text = link.get_text(strip=True)

    if href:
        print(f"{text}: {href}")
```
