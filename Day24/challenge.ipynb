Extend the project to:

Count students per department.
Display the average CGPA for each department

```python
import pandas as pd
students = pd.read_csv("students1.csv")


top = students.sort_values(by="CGPA",ascending=False)
print("Top Students:\n",top)
top.to_csv("Top Students.csv",index=False)
print("\n Cyber Security Students:")
print(students[students["Department"]=="Cyber Security"])
count = students["Department"].value_counts()
print("Count:",count)
average_cgpa = students.groupby("Department")["CGPA"].mean()
print(average_cgpa)
```

```python
import pandas as pd

logs = pd.read_csv("login_logs.csv")

print("Successful Logins:")
print(logs[logs["Status"] == "Success"])
count = logs["Status"].value_counts()
print("Count of Logins:",count)
print("\nFailed Logins:")
print(logs[logs["Status"] == "Failed"])
```
