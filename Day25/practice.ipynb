Exercise 2

Create a table:

students

Columns:

ID
Name
Branch
CGPA

```python
import sqlite3
connectt = sqlite3.connect("college.db")
cursor = connectt.cursor()
cursor.execute("""
              create table if not exists college(
                  id integer,
                  name text,
                  branch text,
                  cgpa real)
                  """)
connectt.commit()
connectt.close()
```

Exercise 4

Display all Cyber Security students

```python
import sqlite3
connectt = sqlite3.connect("college.db")
cursor = connectt.cursor()
cursor.execute("select * from college where branch=?",("CSC",))
rows = cursor.fetchall()
print("CSC Students are:")
for row in rows:
    print(row)
connectt.close()
```
