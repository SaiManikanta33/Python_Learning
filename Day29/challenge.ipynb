🧩 Mini Project
Log File Streamer

Create a file named:

security.log

Example:

INFO User admin logged in
WARNING Failed login for guest
ERROR Database connection failed
INFO User sai logged in



Challenge:
Extend the project to:

Print only ERROR logs.
Count WARNING entries.
Save all ERROR logs to errors.log.
Count the total number of log entries

```python
def process_logs(input_file, error_file):
    warning_count = 0
    total_logs = 0

    with open(input_file, "r") as file, open(error_file, "w") as errors:
        for line in file:
            total_logs += 1
            line = line.strip()

            if "WARNING" in line:
                warning_count += 1

            if "ERROR" in line:
                print(line)
                errors.write(line + "\n")

    return warning_count, total_logs


warnings, total = process_logs("security.log", "errors.log")

print("WARNING entries:", warnings)
print("Total log entries:", total)
```

🌟 Bonus Challenge

Build a live log monitor similar to the Linux tail -f command.

Requirements:

Read new log lines as they are added.
Print them immediately.
Highlight ERROR and CRITICAL messages.

This introduces the concept of real-time log monitoring used in SOC environments.

```python
import time

def monitor_log(file_path):
    with open(file_path, "r") as file:
        file.seek(0, 1)  # Start at the end of the file

        while True:
            line = file.readline()

            if line:
                line = line.strip()

                if "CRITICAL" in line:
                    print(f"\033[95m{line}\033[0m")  # Magenta
                elif "ERROR" in line:
                    print(f"\033[91m{line}\033[0m")  # Red
                else:
                    print(line)
            else:
                time.sleep(0.5)

monitor_log("student_app.log")
```
