Function Execution Logger

```python
from datetime import datetime
def log_execution(func):
    def wrapper(*args,**kwargs):
        print(f"[{datetime.now()}] Running {func.__name__}")
        res =func(*args,**kwargs)
        print(f"[{datetime.now()}] Finished {func.__name__}")
        return res
    return wrapper
@log_execution
def scan():
    print("Scanning network...")
scan()
```

Login Attempt Logger

```python
import time
def audit_log(func):
    def wrapper(name):
        print(f"[{datetime.now()}] Login attempt by {name}")
        return func(name)
    return wrapper
@audit_log
def login(name):
    print(f"Authenticating {name}...")
login("admin")
```
